```python
import secrets, hashlib
```

```python
ALPHABET = "0123456789ABCDEFGHJKMNPQRSTVWXYZ"

def to_crockford_base32(data: bytes) -> str:
    bits = 0
    value = 0
    out = []
    for b in data:
        value = (value << 8) | b
        bits += 8
        while bits >= 5:
            out.append(ALPHABET[(value >> (bits - 5)) & 31])
            bits -= 5
    if bits:
        out.append(ALPHABET[(value << (5 - bits)) & 31])
    return ''.join(out)

def generate_invite_code(groups=2, group_len=4) -> str:
    raw = secrets.token_bytes(16)  # 128-bit
    base32 = to_crockford_base32(raw)
    token = base32[:groups * group_len]
    return '-'.join(token[i:i+group_len] for i in range(0, len(token), group_len))

def hash_invite(token: str) -> str:
    return hashlib.sha256(token.encode('utf-8')).hexdigest()
```
